# Meeting notes summarizer and AP taker#

Python program that takes an transcribed file of a recorded meeting via Google meet or Zoom and generates a consise summary of minutes of the meeting and list of action points.  Based on OpenAI API functionality like  whisper and chat completion.

```python
from google.colab import userdata
import openai
import os

# Step 0: Set your OpenAI API key
secret_key=userdata.get('OPENAI_API_KEY')
openai.api_key = secret_key


# Step 1: Convert txt file into a string variable
def convert_txt_to_string(file_path):
    with open(file_path, 'r') as file:
        output = file.read()
    return output


# Step 2: Summarize the transcription
def summarize_text(transcript: str) -> str:
    prompt = f"Summarize the following transcript into a short meeting summary:\n\n{transcript}"
    response = openai.chat.completions.create(
        model="gpt-4",
        messages=[{"role": "user", "content": prompt}],
        temperature=0.5
    )
    return response.choices[0].message.content.strip()


# Step 3: Extract structured action items
def extract_action_items(summary: str) -> str:
    prompt = f"""Based on the following meeting summary, extract clear, bullet-point action items:\n\n{summary}\n\nFormat like:\n- [Task] — [Owner] — [Deadline if available]"""
    response = openai.chat.completions.create(
        model="gpt-4",
        messages=[{"role": "user", "content": prompt}],
        temperature=0.5
    )
    return response.choices[0].message.content.strip()


# Full pipeline function
def process_meeting_transcription(file_path: str):
    print("\n Converting txt file to string variable...")
    transcript = convert_txt_to_string(file_path)
    print("\n📝 Summary:")
    summary = summarize_text(transcript)
    print(summary)

    print("\n✅ Action Items:")
    action_items = extract_action_items(summary)
    print(action_items)

    # save summarized meeting notes to a txt. file
    os.chdir('/content/sample_data') # replace with your preferred directory to save file
    with open("meeting_summary_aps.txt", "w") as file:
    # Write each string to the file, adding a newline character ('\n') for separation
      file.write("Summary:" + "\n" + summary + "\n")
      file.write("\n" + "Action items:" + "\n" + action_items + "\n")


# Example usage
if __name__ == "__main__":
    process_meeting_transcription("/content/sample_data/SampleTranscript.txt")  # replace with your transcription file path
```
